## 2. Tema 1: Estructuras de Datos Lineales (Pilas y Colas)

Las Pilas siguen la filosofía **LIFO** (último en entrar, primero en salir) y las Colas siguen **FIFO** (primero en entrar, primero en salir).

### Ejercicio 1 (Resuelto): Implementación de una Pila Simple
A continuación se presenta una implementación básica de una Pila utilizando listas de Python.

```python
from typing import Any, List

class Pila:
    def __init__(self) -> None:
        self.items: List[Any] = []

    def push(self, item: Any) -> None:
        self.items.append(item)  # Agrega al final

    def pop(self) -> Any:
        if self.is_empty():
            return None
        return self.items.pop()  # Elimina y retorna el último

    def peek(self) -> Any:
        if self.is_empty():
            return None
        return self.items[-1]

    def is_empty(self) -> bool:
        return len(self.items) == 0

# Probar Pila
stack = Pila()
stack.push(1)
stack.push(2)
print(f"Cima de la pila: {stack.peek()}") # Debe ser 2
print(f"Sacar elemento: {stack.pop()}")    # Saca el 2
```

### Ejercicio 5 (Resuelto - Búsqueda de Error): Optimización de Búsqueda de Duplicados

**Código Erróneo (Bug de Lógica y Eficiencia):** Ejecuta la siguiente función. Tiene un error en los índices de comparación que provoca falsos positivos y además es ineficiente en tiempo.

```python
def tiene_duplicados_buggy(lista: List[int]) -> bool:
    n = len(lista)
    for i in range(n):
        # BUG DE LÓGICA: El bucle interno empieza en 0. 
        # Compara el elemento con sí mismo en la posición i == j y retorna True incorrectamente.
        # BUG DE COMPLEJIDAD: Dos bucles anidados darían O(n^2) de todas formas.
        for j in range(0, n):
            if lista[i] == lista[j]:
                return True
    return False

# --- EJECUTAR CÓDIGO ERRONEO ---
try:
    print("Probando lista sin duplicados [1, 2, 3, 4]...")
    resultado = tiene_duplicados_buggy([1, 2, 3, 4])
    print(f"Resultado obtenido: {resultado}")
    assert resultado == False, "¡Error! Retornó True para una lista limpia."
except AssertionError as e:
    print(f"PRUEBA FALLIDA (Comportamiento esperado): {e}")
```

**¿Por qué está mal?**
1. **Error lógico**: Al evaluar `lista[i] == lista[j]` cuando `i == j`, siempre dará verdadero. Se debería empezar el bucle secundario en `i + 1`.
2. **Error de rendimiento**: El doble bucle anidado realiza $\mathcal{O}(n^2)$ operaciones. Podemos optimizarlo a complejidad lineal $\mathcal{O}(n)$ usando un conjunto (`set`).

**Código Corregido:** Ejecuta la versión optimizada que resuelve ambos problemas.

```python
def tiene_duplicados(lista: List[int]) -> bool:
    # Un set de Python realiza búsquedas en tiempo promedio O(1)
    # Complejidad temporal final: O(n)
    vistos = set()
    for num in lista:
        if num in vistos:
            return True
        vistos.add(num)
    return False

# --- EJECUTAR CÓDIGO CORREGIDO ---
lista_limpia = [1, 2, 3, 4]
lista_duplicada = [1, 2, 3, 2, 4]
print(f"Lista limpia [1, 2, 3, 4] -> Duplicados: {tiene_duplicados(lista_limpia)}")
print(f"Lista con duplicados [1, 2, 3, 2, 4] -> Duplicados: {tiene_duplicados(lista_duplicada)}")
assert tiene_duplicados(lista_limpia) == False
assert tiene_duplicados(lista_duplicada) == True
print("¡Pruebas asintóticas pasadas con éxito!")
```

## 4. Tema 3: Recursividad y Ecuaciones de Recurrencia

Un algoritmo recursivo requiere de al menos un **caso base** (para terminar la ejecución) y un **caso recursivo**.

### Ejercicio 7 (Resuelto): Factorial Recursivo
Cálculo clásico de factorial en donde definimos claramente el caso base para detener las llamadas.

```python
def factorial(n: int) -> int:
    # Caso Base
    if n <= 1:
        return 1
    # Caso Recursivo
    return n * factorial(n - 1)

print(f"Factorial de 5: {factorial(5)}") # 120
assert factorial(5) == 120, "Error en factorial"
```

### Ejercicio 10 (Resuelto): Búsqueda Lineal Simple
Recorre la lista secuencialmente buscando el elemento objetivo. Su complejidad es $\mathcal{O}(n)$.

```python
def busqueda_lineal(lista: List[int], target: int) -> int:
    for i in range(len(lista)):
        if lista[i] == target:
            return i # Retorna el índice si se encuentra
    return -1 # Retorna -1 si no existe

mi_lista = [5, 3, 8, 2, 9]
print(f"Buscar 8 en lista: Índice {busqueda_lineal(mi_lista, 8)}") # 2
assert busqueda_lineal(mi_lista, 8) == 2, "Error en búsqueda lineal"
```

### Ejercicio 11 (Resuelto - Búsqueda de Error): Bucle Infinito en Búsqueda Binaria

**Código Erróneo:** Ejecuta el siguiente algoritmo de búsqueda binaria. Para prevenir congelar tu cuaderno de Jupyter por un bucle infinito, hemos añadido un límite de 30 iteraciones que disparará un error controlado.

```python
def busqueda_binaria_buggy(arr: List[int], target: int) -> int:
    low, high = 0, len(arr) - 1
    iteraciones = 0
    while low <= high:
        iteraciones += 1
        if iteraciones > 30:
            # Salvaguarda para evitar congelar el Kernel de Jupyter
            raise RuntimeError("¡Bucle infinito detectado! La búsqueda no converge.")
            
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            # BUG CRÍTICO: No avanzar el puntero low más allá de mid (low = mid en vez de mid + 1).
            # Si low y high están a 1 elemento de distancia, se quedará aquí para siempre.
            low = mid
        else:
            high = mid - 1
    return -1

# --- EJECUTAR CÓDIGO ERRONEO ---
try:
    print("Buscando el número 4 en la lista ordenada [1, 3, 5]...")
    busqueda_binaria_buggy([1, 3, 5], 4)
except Exception as e:
    print(f"ERROR DETECTADO: {type(e).__name__} -> {e}")
```

**¿Por qué está mal?**
Si la mitad evaluada es menor al `target`, la búsqueda debe reanudarse estrictamente en la siguiente mitad, actualizando `low = mid + 1`. Al asignar simplemente `low = mid`, cuando los punteros se acercan y `low == mid` en división entera (por ejemplo `low=1, high=2, mid=1`), el valor de `low` nunca avanza, creando un ciclo sin fin.

**Código Corregido:** Ejecuta la versión correcta que avanza los punteros de forma estricta.

```python
def busqueda_binaria(arr: List[int], target: int) -> int:
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1  # Incremento correcto
        else:
            high = mid - 1  # Decremento correcto
    return -1

# --- EJECUTAR CÓDIGO CORREGIDO ---
lista_ordenada = [1, 3, 5, 7, 9, 11]
idx = busqueda_binaria(lista_ordenada, 7)
print(f"Buscar 7 en {lista_ordenada} -> Encontrado en índice: {idx}")
assert idx == 3, "Error en índice de búsqueda binaria"
print("¡Búsqueda binaria terminada con éxito sin bucles infinitos!")
```
