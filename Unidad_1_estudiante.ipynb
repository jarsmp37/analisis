## 2. Tema 1: Estructuras de Datos Lineales (Pilas y Colas)

Las Pilas siguen la filosofía **LIFO** (último en entrar, primero en salir) y las Colas siguen **FIFO** (primero en entrar, primero en salir).

### Ejercicio 1 (Resuelto): Implementación de una Pila Simple
A continuación se presenta una implementación básica de una Pila utilizando listas de Python.

```python
from typing import Any, List

class Pila:
    def __init__(self) -> None:
        self.items: List[Any] = []

    def push(self, item: Any) -> None:
        self.items.append(item)  # Agrega al final

    def pop(self) -> Any:
        if self.is_empty():
            return None
        return self.items.pop()  # Elimina y retorna el último

    def peek(self) -> Any:
        if self.is_empty():
            return None
        return self.items[-1]

    def is_empty(self) -> bool:
        return len(self.items) == 0

# Probar Pila
stack = Pila()
stack.push(1)
stack.push(2)
print(f"Cima de la pila: {stack.peek()}") # Debe ser 2
print(f"Sacar elemento: {stack.pop()}")    # Saca el 2
```

## 3. Tema 2: Análisis Asintótico y Notación Big-O

La notación Big-O ($\mathcal{O}$) describe el límite superior del tiempo de ejecución de un algoritmo en el peor escenario asintótico.

### Ejercicio 4 (Resuelto): Complejidad de Bucle Simple

### Ejercicio 5 (Resuelto - Búsqueda de Error): Optimización de Búsqueda de Duplicados

**Código Erróneo:** Ejecuta el siguiente código. Funciona bien, pero tiene una complejidad cuadrática de $\mathcal{O}(n^2)$ debido a los bucles anidados. Busca optimizarlo a $\mathcal{O}(n)$ usando una estructura auxiliar (tabla hash / conjunto).

```python
def tiene_duplicados_lento(lista: List[int]) -> bool:
    n = len(lista)
    for i in range(n):
        for j in range(i + 1, n):
            if lista[i] == lista[j]:
                return True
    return False

# --- EJECUTAR COMPROBACIÓN ---
arr = list(range(1000))
import time
t0 = time.time()
tiene_duplicados_lento(arr)
t1 = time.time()
print(f"Algoritmo O(N^2) tardó: {(t1 - t0)*1000:.3f} ms")
```

**¿Por qué está mal?**
Los bucles aninados obligan a comparar todos los elementos entre sí, lo que requiere tiempo cuadrático. Podemos utilizar un conjunto (`set` en Python) para almacenar los elementos ya visitados y comprobar en tiempo constante $\mathcal{O}(1)$ si un elemento ya fue visto, reduciendo el algoritmo a tiempo lineal $\mathcal{O}(n)$.

```python
def tiene_duplicados_rapido(lista: List[int]) -> bool:
    vistos = set()
    for num in lista:
        if num in vistos:  # O(1)
            return True
        vistos.add(num)    # O(1)
    return False

# --- EJECUTAR CÓDIGO OPTIMIZADO ---
arr = list(range(1000))
t0 = time.time()
tiene_duplicados_rapido(arr)
t1 = time.time()
print(f"Algoritmo O(N) tardó: {(t1 - t0)*1000:.3f} ms")
assert tiene_duplicados_rapido([1, 2, 3, 2]) == True
assert tiene_duplicados_rapido([1, 2, 3, 4]) == False
print("¡Prueba pasada con éxito!")
```

## 4. Tema 3: Recursividad y Ecuaciones de Recurrencia

La recursividad es una técnica donde una función se llama a sí misma para resolver subproblemas más pequeños de su misma naturaleza. Requiere siempre un **Caso Base** para detenerse.

### Ejercicio 7 (Resuelto): Factorial Recursivo
Ejemplo clásico de recursión que calcula el factorial de un número $N$.

```python
def factorial(n: int) -> int:
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(f"Factorial de 5: {factorial(5)}")
```

### Ejercicio 13 (Resuelto): Algoritmo de Euclides para el MCD
Este algoritmo calcula el Máximo Común Divisor (MCD) de dos números enteros restando consecutivamente el menor del mayor hasta que ambos sean iguales.

```python
def mcd_euclides(a: int, b: int) -> int:
    iteraciones = 0
    while a != b:
        iteraciones += 1
        if a > b:
            a = a - b
        else:
            b = b - a
    print(f"[INFO] MCD encontrado en {iteraciones} iteraciones.")
    return a

print(f"MCD de 24 y 18: {mcd_euclides(24, 18)}")
```

## 7. Tema 6: Cota Inferior Asintótica (Notación Omega $\Omega$) (Syllabus 1.2 - Parte 1)

**Explicación en Lenguaje Natural:**
Mientras que Big-O nos indica lo *peor* que puede comportarse un algoritmo, la notación **Omega ($\\Omega$)** nos indica lo *mejor*. Representa la cota inferior asintótica: nos asegura que el algoritmo tardará **al menos** esa cantidad de tiempo o pasos, sin importar cuán favorables sean los datos de entrada.

Por ejemplo, si estás buscando un nombre en una lista de contactos, el **mejor escenario** es que sea el primer contacto que revisas. Ese mejor caso se representa como una cota inferior constante $\Omega(1)$, ya que siempre tomará como mínimo 1 comparación.

### Ejercicio 16 (Resuelto): Encontrar el Primer Número Par
Este algoritmo busca el primer número par en un arreglo. Analizaremos su cota inferior $\Omega$ y superior $O$.

```python
def encontrar_primer_par(lista: List[int]) -> int:
    for num in lista:
        if num % 2 == 0:
            return num
    return -1

print(f"Mejor caso (Omega): {encontrar_primer_par([2, 5, 7, 9])}")
print(f"Peor caso (Big-O): {encontrar_primer_par([1, 3, 5, 8])}")
```

## 8. Tema 7: Cota Ajustada Asintótica (Notación Theta $\Theta$) (Syllabus 1.2 - Parte 2)

**Explicación en Lenguaje Natural:**
La notación **Theta ($\Theta$)** representa la cota ajustada. Decimos que un algoritmo es $\Theta(g(n))$ si su tiempo de ejecución crece a la misma velocidad tanto en el mejor como en el peor de los casos. En términos informales, significa que el algoritmo no tiene "atajos" ni sorpresas: siempre tarda proporcionalmente lo mismo sin importar la distribución de la entrada.

### Ejercicio 19 (Resuelto): Sumar una Matriz Bidimensional cuadrática
Este algoritmo suma todos los valores de una matriz cuadrática de $N \times N$. Como está obligado a recorrer todas las celdas, su cota ajustada es cuadrática $\Theta(N^2)$.

```python
def sumar_matriz(matriz: List[List[int]]) -> int:
    total = 0
    n = len(matriz)
    for i in range(n):
        for j in range(n):
            total += matriz[i][j]
    return total

m = [[1, 2], [3, 4]]
print(f"Suma total de matriz: {sumar_matriz(m)}")
```

## 9. Tema 8: Análisis de Operaciones Elementales (Syllabus 1.3.1)

**Explicación en Lenguaje Natural:**
Para medir con precisión matemática el tiempo de ejecución de un programa, contamos la cantidad exacta de **operaciones elementales** que realiza el procesador. Una operación elemental es cualquier acción básica e indivisible, tal como:
- Asignar un valor a una variable (`x = 5`).
- Comparar dos valores en una condición (`x > 10`).
- Hacer una operación aritmética (`a + b`).
- Leer o escribir en un arreglo o retornar un valor.

Sumando la cantidad de veces que se ejecutan estas operaciones en el código en función del tamaño de entrada $N$, podemos modelar matemáticamente el rendimiento exacto de nuestro software.

### Ejercicio 23 (Resuelto - Búsqueda de Error): Conteo Erróneo en Bucle Acumulador

**Código Erróneo:** Ejecuta el siguiente código. El estudiante afirma: *'Como la función tiene un bucle simple de rango N, realiza exactamente N operaciones elementales.'* Analiza por qué este conteo es erróneo y subestima la realidad.

```python
def acumular_valores(n: int) -> int:
    suma = 0
    for i in range(n):
        suma += i
    return suma

print(f"Resultado: {acumular_valores(3)}")
```

**¿Por qué está mal el análisis?**
El estudiante confunde la cantidad de *iteraciones* de un bucle con la cantidad de *operaciones elementales*. En cada iteración del bucle, el computador realiza la comparación (`i < n`), la suma (`suma + i`), la asignación (`suma = ...`), y el incremento del índice (`i = i + 1`). Por lo tanto, realiza unas 4 operaciones por iteración, más las de inicio y retorno. El total real es $4N + 3$ operaciones.

**Código Corregido:** Ejecuta la simulación con conteo explícito.

```python
def acumular_valores_con_conteo(n: int) -> (int, int):
    operaciones = 0
    suma = 0
    operaciones += 1 # Asignación suma = 0
    i = 0
    operaciones += 1 # Asignación i = 0
    while i < n:
        operaciones += 1 # Comparación i < n
        suma += i
        operaciones += 2 # Suma y asignación
        i += 1
        operaciones += 2 # Incremento y asignación
    operaciones += 1 # Comparación fallida
    operaciones += 1 # Retorno
    return suma, operaciones

res, ops = acumular_valores_con_conteo(3)
print(f"Para N=3: Resultado={res}, Operaciones={ops}")
```

## 10. Tema 9: Análisis de Estructuras de Control (Syllabus 1.3.2)

**Explicación en Lenguaje Natural:**
Las **estructuras de control** (bucles `for`/`while`, decisiones `if`/`else`) determinan el flujo de ejecución del programa. Para analizarlas:
- **Bucles independientes:** Si tenemos dos bucles secuenciales, sus tiempos de ejecución se **suman** ($N + N = 2N \implies \mathcal{O}(N)$).
- **Bucles anidados independientes:** Si un bucle de tamaño $N$ está dentro de otro de tamaño $N$, sus iteraciones se **multiplican** ($N \times N = N^2 \implies \mathcal{O}(N^2)$).
- **Bucles anidados dependientes:** Ocurren cuando el rango del bucle interno cambia en cada paso según el índice del bucle externo (por ejemplo, ir de `i` hasta `N`). Aquí, la cantidad de iteraciones no es un producto directo, sino la suma aritmética de una progresión.

### Ejercicio 25 (Resuelto): Bucles Independientes vs Anizados
Veremos cómo se calcula el número de pasos en dos bucles secuenciales contra dos bucles anidados.

```python
def ejecutar_estructuras(N: int):
    pasos_suma = 0
    pasos_mult = 0
    for i in range(N):
        pasos_suma += 1
    for j in range(N):
        pasos_suma += 1
    for i in range(N):
        for j in range(N):
            pasos_mult += 1
    print(f"Para N={N} -> Pasos Secuenciales: {pasos_suma}, Pasos Anidados: {pasos_mult}")

ejecutar_estructuras(5)
```

## 11. Tema 10: Interpretación Algorítmica: Mejor, Peor y Caso Promedio (Syllabus 1.3.3)

**Explicación Detallada en Lenguaje Natural (Sin Matemáticas Abstractas):**

El tiempo que tarda un programa en ejecutarse casi nunca depende únicamente del *tamaño* de la entrada ($N$). También depende del *contenido* y del *orden* en el que se presentan esos datos. Para entender el comportamiento completo de un algoritmo, los ingenieros dividen el análisis en tres casos principales:

### 1. El Mejor Caso (Cota Inferior / Optimista)
**La Analogía de las Llaves:** Imagina que tienes una mochila llena de objetos desordenados y necesitas encontrar tus llaves. El **mejor caso** ocurre si abres la cremallera, introduces la mano por primera vez y, por pura fortuna, lo primero que tocan tus dedos son las llaves. ¡Problema resuelto en 1 paso!

En computación, el mejor caso representa la distribución de datos más favorable posible para el algoritmo. Por ejemplo, si buscas un valor en una lista desordenada y este valor resulta ser el que está en la primera posición. El programa realiza una sola comparación y termina de inmediato. Se denota generalmente usando la notación **Omega ($\\Omega(1)$)**.

### 2. El Peor Caso (Cota Superior / Garantía de Tiempo Máximo)
**La Analogía de las Llaves:** El **peor caso** ocurre cuando las llaves están en el fondo de la mochila, sepultadas bajo cuadernos, cables y estuches, o peor aún: ¡cuando no están en la mochila! En este escenario, estás obligado a sacar y revisar cada objeto uno por uno hasta llegar al final o comprobar que no está.

En el software, el peor caso representa la distribución de datos más hostil y desfavorable. Por ejemplo, buscar un elemento que está en la última posición de la lista o que no existe. Estás obligado a recorrer los $N$ elementos completos antes de poder dar una respuesta. Es la métrica más importante en ingeniería porque representa una **garantía de tiempo límite**: nos asegura que el programa nunca tardará más de eso. Se representa con la notación **Big-O ($O(N)$)**.

### 3. El Caso Promedio (Rendimiento Esperado en la Vida Real)
**La Analogía de las Llaves:** En tu día a día, no vas a encontrar las llaves al primer intento (mejor caso) ni vas a tener que vaciar toda la mochila en cada ocasión (peor caso). Por lo general, las llaves estarán en algún punto intermedio. Si buscas tus llaves cien veces en días diferentes, en promedio las encontrarás a la mitad del camino.

El **caso promedio** modela el comportamiento del algoritmo frente a datos reales, aleatorios o distribuidos uniformemente. Si buscas un elemento aleatorio en una lista de tamaño $N$, en promedio realizarás $N/2$ búsquedas. Sigue creciendo linealmente ($O(N)$), pero nos da una expectativa realista y práctica de cómo se comportará nuestra aplicación en producción bajo condiciones de uso ordinarias.

### Ejercicio 28 (Resuelto): Simulación Interactiva de Casos en Búsqueda Lineal
Ejecuta el código de abajo. Analizaremos cómo se comporta el conteo de operaciones elementales reales según la posición del elemento objetivo.

```python
def buscar_con_contador(lista: List[int], objetivo: int) -> (int, int):
    ops = 0
    n = len(lista)
    ops += 1
    for i in range(n):
        ops += 2 # Control de bucle y comparación
        if lista[i] == objetivo:
            ops += 1
            return i, ops
    ops += 1
    return -1, ops

arr = [10, 20, 30, 40, 50]
pos, ops_mejor = buscar_con_contador(arr, 10)
print(f"[MEJOR CASO]  Buscar 10: posición={pos}, operaciones={ops_mejor}")
pos, ops_peor = buscar_con_contador(arr, 99)
print(f"[PEOR CASO]   Buscar 99: posición={pos}, operaciones={ops_peor}")
pos, ops_medio = buscar_con_contador(arr, 30)
print(f"[CASO MEDIO]  Buscar 30: posición={pos}, operaciones={ops_medio}")
```
